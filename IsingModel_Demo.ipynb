```python
import random as rd
import numpy as np
import time as t
import matplotlib.pyplot as plt
import matplotlib.animation as Ani
from IPython.display import display, HTML
SIZE = 10
dTemperature = 1.0
J = 1
B = 1
#配色白為能量+1，黑為能量-1
```

```python
def CalculateEnergy(iSpin2D):
    iE_Total = 0
    iNeighbors = 0
    for i in range(SIZE):
        for j in range(SIZE):
            iNeighbors = (iSpin2D[(i - 1 + SIZE) % SIZE][j] 
                        + iSpin2D[(i + 1) % SIZE][j]
                        + iSpin2D[i][(j - 1 + SIZE) % SIZE]
                        + iSpin2D[i][(j + 1) % SIZE])
            iE_Total = iE_Total - J * iSpin2D[i][j] * iNeighbors
    return (iE_Total / 2)

def PrintSpin(iSpin2D):
    for i in range(SIZE):
        for j in range(SIZE):
            print(f"{iSpin2D[i][j]:2d}", end=" ")
        print()
def CreateAnimation(spin_history, interval=80):
    fig, ax = plt.subplots()
    im = ax.imshow(spin_history[0],
                   cmap="binary",
                   vmin=-1,
                   vmax=1
                    )
    ax.set_title("Animation")
    ax.set_xticks([])
    ax.set_yticks([])

    def update(frame):
        im.set_array(spin_history[frame])
        ax.set_xlabel(f"Frame {frame + 1}/{len(spin_history)}")
        return [im]

    ani = Ani.FuncAnimation(
        fig, update, frames=len(spin_history), interval=interval, blit=True
    )
    plt.close(fig)
    return ani
```

```python
iEnergy_list  = []
iSpin_History = []

def main(iInterval_Record=50): 
    global iEnergy_list, iSpin_History
    iEnergy_list  = []
    iSpin_History = []
    
    iSpin2D = [[0]*SIZE for _ in range(SIZE)]
    dProb_Target = [0.0] * 3
    rand = rd.random()
    iStep_Max = 5000
    iE_Total = 0
    iX = 0 
    iY = 0
    iE_Delta = 0
    dProb = 0
    iSpin_History.append(np.array(iSpin2D).copy())

    for i in range(SIZE):
        for j in range(SIZE):
            r = rd.randrange(2)      # 0 或 1
            iSpin2D[i][j] = r * 2 - 1

    dProb_Target[0] = 1.0;
    dProb_Target[1] = np.exp(-4.0 * J / dTemperature)
    dProb_Target[2] = np.exp(-8.0 * J / dTemperature)

    PrintSpin(iSpin2D)
    iE_Total = CalculateEnergy(iSpin2D)
    print("Initial energy = ", iE_Total)
    for iStep in range(1 , iStep_Max):
        iEnergy_list.append(iE_Total)
        iX = rd.randrange(SIZE)
        iY = rd.randrange(SIZE)
        iNeighbors = (iSpin2D[(iX - 1 + SIZE) % SIZE][iY] 
                    + iSpin2D[(iX + 1) % SIZE][iY]
                    + iSpin2D[iX][(iY - 1 + SIZE) % SIZE]
                    + iSpin2D[iX][(iY + 1) % SIZE])
        iE_Delta = 2 * J * iSpin2D[iX][iY] * iNeighbors
        
        if(iE_Delta <= 0):
            iE_Total += iE_Delta
            iSpin2D[iX][iY] = -iSpin2D[iX][iY]
        else:
            dProb = rd.random()
            dIndex = iE_Delta / (4 * J)
            if(dProb < dProb_Target[int(dIndex)]):
                iE_Total += iE_Delta
                iSpin2D[iX][iY] = -iSpin2D[iX][iY]

        if(iStep % iInterval_Record == 0):
            iSpin_History.append(np.array(iSpin2D).copy())

    iEnergy_list.append(iE_Total)
    print("Final energy = ", iE_Total)
    PrintSpin(iSpin2D)
    #iE_Total = CalculateEnergy(iSpin2D)
    #print("Final energy = ", iE_Total)    
    return iEnergy_list, iSpin_History
if __name__ == "__main__":
    iEnergy_list, iSpin_History = main(iInterval_Record=50)

    ani = CreateAnimation(iSpin_History, interval=80)
    display(HTML(ani.to_jshtml()))
```
